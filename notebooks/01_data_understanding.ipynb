### Phase 1 — Initial Findings

- Dataset contains 1067371 rows and 8 columns.
- 2 columns contain missing values.
- 34335 duplicate rows were found.
- Dataset covers the period from 2009-12-01 to 2011-12-09.
- There are 47635 unique invoices.
- There are 5305 unique products.
- There are 5942 unique customers.
- There are 43 countries.
- Quantity contains 22950 negative values and 0 'zero' values.
- Price contains 6207 zero/negative values.
- 19494 transaction records have invoice numbers beginning with `C` and may represent cancellations; this will be investigated during data cleaning.
- Other suspicious observations: `InvoiceDate` is stored as an object, `Customer ID` is stored as float64, and several invoice/product codes
  require further investigation before cleaning..
